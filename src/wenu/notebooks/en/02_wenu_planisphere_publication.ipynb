# Publication planisphere

This version uses current Wenu public composition. It retains La Ligua, the local date, east-west reversal and the example magnitude limit. The former renderer calls are replaced by atlas style and canonical export; visual identity with the obsolete notebook is not claimed.


```python
import argparse
from pathlib import Path

from wenu import (
    Observer, add_chart_cli_arguments, draw_chart_view_from_arguments,
    generate_celestial_sphere, get_chart_view,
)
```

## Editable parameters


```python
LOCAL_TIME = "2026-08-15 21:00"
LOCATION = "La Ligua"
MAGNITUDE_LIMIT = 6.0
FLIP_EAST_WEST = True
LANGUAGE = 'en'
OUTPUT_FILE = Path('carta_33S_2026-08-15_21h_explicit_wenu.png')
TITLE = 'Southern sky — La Ligua — 15 August 2026, 21:00'
```

Lines, labels, boundaries, planes and poles are independent requests. Geometry, transforms and export belong to Wenu.


```python
# Explicit requests use the same controls as the six canonical examples.
parser = add_chart_cli_arguments(argparse.ArgumentParser(), default_output=OUTPUT_FILE)
arguments = parser.parse_args([
    "--style", "atlas", "--mode", "print",
    "--magnitude-limit", str(MAGNITUDE_LIMIT),
    "--constellation-lines", "all", "--constellation-labels", "all",
    "--constellation-boundaries", "all",
    "--grid-references", "all", "--poles", "--pole-labels",
    "--horizon", "--output", str(OUTPUT_FILE),
])
```

## Generate and export

The observer also closes if generation fails. Open the reported PNG to view the chart. Rerun from the first cell after changing parameters.


```python
observer = Observer(location=LOCATION, time=LOCAL_TIME)
try:
    sky = generate_celestial_sphere()
    view = get_chart_view(
        sky, observer, family="planisphere", projection="stereographic",
        mirror_ew=FLIP_EAST_WEST,
    )
    results = draw_chart_view_from_arguments(
        view, arguments, stem=OUTPUT_FILE.stem, title=TITLE, language=LANGUAGE,
    )
finally:
    observer.close()

for result in results:
    print(result.output)
```
